```python

import numpy as np

class SVM:

    def __init__(self, lr=0.001, n=1000):

        self.lr = lr
        self.n = n
        self.w = None
        self.b = None

    def fit(self, X, y):

        n_samples, n_features = X.shape

        y = np.where(y <= 0, -1, 1)

        self.w = np.zeros(n_features)
        self.b = 0.0

        for _ in range(self.n):

            scores = X @ self.w + self.b

            margins = y * scores

            # Find points violating hard margin constraint
            active = margins < 1

            dw = self.w
            db = 0.0

            if np.any(active):

                dw = dw - np.sum(
                    X[active] * y[active, np.newaxis],
                    axis=0
                )

                db = db - np.sum(y[active])

            self.w -= self.lr * dw
            self.b -= self.lr * db

        return self

    def predict(self, X):

        scores = X @ self.w + self.b

        return np.where(scores >= 0, 1, -1)
```
