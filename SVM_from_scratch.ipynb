```python
import numpy as np
```

```python
class SVM:
    def __init__(self,lr=0.001,lmbda=0.01,n=1000):
        self.lr=lr
        self.lmbda=lmbda
        self.n=n
        self.w=None
        self.b=None

    def fit(self,X,y):
        n_samples,n_features=X.shape
        y=np.where(y<=0, -1,1)
        self.w=np.zeros(n_features)
        self.b=0.0

        for _ in range(self.n):
            scores=X@self.w+self.b
            margins=y*scores

            active=margins<1
            dw=self.lmbda*self.w
            db=0.0
        
        if np.any(active):
            dw-=np.sum(X[active]*y[active,np.newaxis],
                       axis=0)/n_samples

            db-=np.sum(y[active])/n_samples

            self.w-=self.lr*dw
            self.b-=self.lr*db
    def predict(self,X):
        scores=X@self.w + self.b
        return np.where(scores >=0 ,1,-1)
    
```

```python
#training data
```

```python
X = np.array([
    [1, 2],
    [2, 3],
    [3, 3],
    [6, 5],
    [7, 7],
    [8, 6]
], dtype=float)

# Labels: 0 and 1 are accepted as input
y = np.array([0, 0, 0, 1, 1, 1])
```

```python
model=SVM(lr=0.001,lmbda=0.1,n=5000)
model.fit(X,y)

predictions=model.predict(X)
predictions
```
