```python

import numpy as np

class MyNaiveBayes:

    def fit(self, X, y):

        # Find unique classes
        self.classes = np.unique(y)

        # Store mean and variance for each class
        self.mean = {}
        self.var = {}
        self.prior = {}

        # Total number of samples
        total_samples = len(y)

        # Calculate values class by class
        for c in self.classes:

            # Select samples belonging to class c
            X_c = X[y == c]

            # Calculate mean of every feature
            self.mean[c] = np.mean(X_c, axis=0)

            # Calculate variance of every feature
            self.var[c] = np.var(X_c, axis=0)

            # Calculate prior probability
            self.prior[c] = len(X_c) / total_samples

        return self

    def calculate_likelihood(self, x, c):

        mean = self.mean[c]
        var = self.var[c]

        # Avoid division by zero
        var = var + 1e-9

        # Gaussian probability
        numerator = np.exp(-((x - mean) ** 2) / (2 * var))

        denominator = np.sqrt(2 * np.pi * var)

        probability = numerator / denominator

        return probability

    def predict(self, X):

        predictions = []

        # Process each test sample
        for x in X:

            probabilities = []

            # Calculate probability for every class
            for c in self.classes:

                likelihood = self.calculate_likelihood(x, c)

                # Log probability to avoid numerical underflow
                log_probability = np.log(self.prior[c])

                log_probability += np.sum(np.log(likelihood))

                probabilities.append(log_probability)

            # Select class having maximum probability
            best_class = self.classes[np.argmax(probabilities)]

            predictions.append(best_class)

        return np.array(predictions)
```
