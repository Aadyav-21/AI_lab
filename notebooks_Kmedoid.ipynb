```python
import numpy as np
import pandas as pd
```

```python
class KMedoids:

    def __init__(self, k, max_iter=100):
        self.k = k
        self.max_iter = max_iter

    # Euclidean distance
    def distance(self, a, b):
        return np.sqrt(np.sum((a - b) ** 2))

    def fit(self, X):

        n = len(X)

        # Step 1: Select random medoids
        medoids = np.random.choice(n, self.k, replace=False)

        for it in range(self.max_iter):

            # Step 2: Create empty clusters
            clusters = []

            for i in range(self.k):
                clusters.append([])

            # Step 3: Assign points to nearest medoid
            for i in range(n):

                distances = []

                for j in range(self.k):
                    d = self.distance(X[i], X[medoids[j]])
                    distances.append(d)

                c = np.argmin(distances)
                clusters[c].append(i)

            # Step 4: Find new medoids
            new_medoids = []

            for i in range(self.k):

                cluster = clusters[i]

                if len(cluster) == 0:
                    new_medoids.append(medoids[i])
                    continue

                min_cost = float('inf')
                best = cluster[0]

                for p in cluster:

                    cost = 0

                    for q in cluster:
                        cost = cost + self.distance(X[p], X[q])

                    if cost < min_cost:
                        min_cost = cost
                        best = p

                new_medoids.append(best)

            # Step 5: Check convergence
            if np.array_equal(medoids, new_medoids):
                break

            medoids = np.array(new_medoids)

        self.medoid_indices = medoids
        self.medoids = X[medoids]

        # Step 6: Final cluster labels
        self.labels = np.zeros(n, dtype=int)

        for i in range(n):

            distances = []

            for j in range(self.k):
                d = self.distance(X[i], self.medoids[j])
                distances.append(d)

            self.labels[i] = np.argmin(distances)

        return self


    def predict(self, X):

        labels = []

        for i in range(len(X)):

            distances = []

            for j in range(self.k):
                d = self.distance(X[i], self.medoids[j])
                distances.append(d)

            c = np.argmin(distances)
            labels.append(c)

        return np.array(labels)
```
