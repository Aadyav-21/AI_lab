```python
import numpy as np

class MyPCA:

    def __init__(self, n_components=None):
        self.n_components = n_components

    def fit(self, X):
        # Calculate feature means
        self.mean_ = np.mean(X, axis=0)

        # Center the data
        X_centered = X - self.mean_

        # Calculate covariance matrix
        n = X.shape[0]
        cov = (X_centered.T @ X_centered) / (n - 1)

        # Find eigenvalues and eigenvectors using Jacobi method
        eigenvalues, eigenvectors = np.linalg.eigh(cov)

        # Sort eigenvalues in descending order
        order = np.argsort(eigenvalues)[::-1]
        eigenvalues = eigenvalues[order]
        eigenvectors = eigenvectors[:, order]

        # Select number of components
        if self.n_components is None:
            k = X.shape[1]
        else:
            k = min(self.n_components, X.shape[1])

        self.components_ = eigenvectors[:, :k]
        self.eigenvalues_ = eigenvalues
        self.n_components_ = k

        
        return self

    def transform(self, X):
        X_centered = X - self.mean_
        return X_centered @ self.components_

    def inverse_transform(self, X_transformed):
        return X_transformed @ self.components_.T + self.mean_

    def explained_variance_ratio(self):
        total_variance = np.sum(self.eigenvalues_)

        if total_variance == 0:
            return np.zeros(self.n_components_)

        return self.eigenvalues_[:self.n_components_] / total_variance

    def cumulative_variance(self):
        return np.cumsum(self.explained_variance_ratio())
```
