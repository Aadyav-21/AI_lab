```python
import numpy as np

class KNN:

    def __init__(self, k=3):
        self.k = k

    # Store training data
    def fit(self, X, y):
        self.X_train = X
        self.y_train = y

    # Calculate Euclidean distance
    def distance(self, x1, x2):
        sum = 0

        for i in range(len(x1)):
            sum += (x1[i] - x2[i]) ** 2

        return np.sqrt(sum)

    # Predict class for one data point
    def predict_one(self, x):

        distances = []

        # Calculate distance from every training point
        for i in range(len(self.X_train)):

            d = self.distance(x, self.X_train[i])

            distances.append((d, self.y_train[i]))

        # Sort according to distance
        distances.sort()

        # Select K nearest neighbors
        neighbors = distances[:self.k]

        # Count votes of each class
        votes = {}

        for d, label in neighbors:

            if label in votes:
                votes[label] += 1
            else:
                votes[label] = 1

        # Find class with maximum votes
        ans = max(votes, key=votes.get)

        return ans

    # Predict for multiple data points
    def predict(self, X):

        predictions = []

        for x in X:
            ans = self.predict_one(x)
            predictions.append(ans)

        return np.array(predictions)
```
